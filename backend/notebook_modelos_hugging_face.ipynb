```python
import math
def split_text_into_chunks(text: str, chunk_size: int, overlap: int) -> list[str]:
    """
    Divide un texto largo en fragmentos (chunks) con superposición.

    Args:
        text (str): El texto de entrada largo.
        chunk_size (int): El tamaño máximo deseado para cada fragmento (en caracteres).
        overlap (int): El número de caracteres a solapar entre fragmentos consecutivos.

    Returns:
        list[str]: Una lista de los fragmentos de texto resultantes.
    """
    if chunk_size <= overlap:
        raise ValueError("chunk_size debe ser mayor que overlap")

    # Calcular el paso para el siguiente fragmento (chunk_size - overlap)
    step = chunk_size - overlap

    # Calcular el número total de fragmentos necesarios
    # Usamos math.ceil para asegurarnos de que el final del texto no se pierda
    num_chunks = math.ceil(len(text) / step) if len(text) > chunk_size else 1

    chunks = []
    for i in range(num_chunks):
        start_index = i * step
        end_index = start_index + chunk_size

        # Asegurarse de no exceder la longitud del texto
        chunk = text[start_index:end_index]
        chunks.append(chunk)

    return chunks
```

```python
import io
import PyPDF2

def extract_text_from_pdf_bytes(pdf_bytes: bytes) -> str:
    """
    Extrae el texto de un archivo PDF a partir de su contenido en bytes.
    """
    full_text = ""
    with io.BytesIO(pdf_bytes) as pdf_file:
        reader = PyPDF2.PdfReader(pdf_file)
        for page in reader.pages:
            text = page.extract_text()
            if text:
                full_text += text + "\n"
    return full_text

```

```python
def recursive_summarize(text: str, chunk_size: int = 1000, overlap: int = 150) -> str:
    """
    Resume un texto largo de forma recursiva.
    """

    # Dividir el texto en fragmentos
    chunks = split_text_into_chunks(text, chunk_size, overlap)

    # Resumir cada fragmento individualmente
    summaries = [summarizer(chunk, max_length=150, min_length=40, do_sample=False)[0]['summary_text'] for chunk in chunks]

    print("LONGITUD SUMMARIES", len(summaries))

    # Combinar los resúmenes parciales
    combined_summary = " ".join(summaries)

    print("TERMINADO COMBINED SUMMARY")
    # Verificar si se necesita otra iteración (recursión)
    # Se estima la longitud en "tokens" de forma aproximada contando palabras.
    if len(combined_summary.split()) > (chunk_size * 0.75): # Umbral heurístico
        print(f"Resumen combinado aún largo ({len(combined_summary.split())} palabras). Recurriendo...")
        return recursive_summarize(combined_summary, chunk_size, overlap)
    else:
        # Resumen final si el texto combinado ya es corto
        print("Resumen finalizando. Generando resumen final.")
        return summarizer(combined_summary, max_length=300, min_length=100, do_sample=False)[0]['summary_text']


final_summary = recursive_summarize(attent0)
```
